```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("earthquake_1995-2023.csv")

print(df.shape)
df.head()
```

```python
country_count = df['country'].value_counts().head(10)

plt.figure(figsize=(10,5))
country_count.plot(kind='bar')

plt.title('Top 10 Countries with Earthquakes')
plt.xlabel('Country')
plt.ylabel('Number of Earthquakes')
plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as colors

country_count = df['country'].value_counts().head(10)

# Create figure and axes
fig, ax = plt.subplots(figsize=(10, 5))

# Normalize earthquake counts
norm = colors.Normalize(
    vmin=country_count.min(),
    vmax=country_count.max()
)

# Create colors based on number of earthquakes
bar_colors = cm.viridis(norm(country_count.values))

# Create bars
bars = ax.bar(
    country_count.index,
    country_count.values,
    color=bar_colors
)

# Titles and labels
ax.set_title('Top 10 Countries with Earthquakes')
ax.set_xlabel('Country')
ax.set_ylabel('Number of Earthquakes')
ax.tick_params(axis='x', rotation=45)

# Display values on top of bars
for bar in bars:
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(int(bar.get_height())),
        ha='center',
        va='bottom'
    )

# Add color map
sm = cm.ScalarMappable(norm=norm, cmap='viridis')
sm.set_array(country_count.values)

fig.colorbar(
    sm,
    ax=ax,
    label='Number of Earthquakes'
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df['magnitude'], bins=10, color='skyblue', edgecolor='black')

plt.title('Distribution of Earthquake Magnitudes')
plt.xlabel('Earthquake Magnitude')
plt.ylabel('Number of Earthquakes')

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Create magnitude groups
df['Magnitude_Group'] = pd.cut(
    df['magnitude'],
    bins=[0, 3, 5, 7, 10],
    labels=['0-3', '3-5', '5-7', '7-10']
)

# Get top 5 countries
top_countries = df['country'].value_counts().head(5).index

grouped_data = pd.crosstab(
    df[df['country'].isin(top_countries)]['country'],
    df[df['country'].isin(top_countries)]['Magnitude_Group']
)

# Plot grouped bar chart
grouped_data.plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Earthquakes by Country and Magnitude')
plt.xlabel('Country')
plt.ylabel('Number of Earthquakes')
plt.xticks(rotation=45)
plt.legend(title='Magnitude')
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Create magnitude groups
df['Magnitude_Group'] = pd.cut(
    df['magnitude'],
    bins=[0, 3, 5, 7, 10],
    labels=['0-3', '3-5', '5-7', '7-10']
)

# Get top 5 countries
top_countries = df['country'].value_counts().head(5).index

# Create data for stacked bar chart
stacked_data = pd.crosstab(
    df[df['country'].isin(top_countries)]['country'],
    df[df['country'].isin(top_countries)]['Magnitude_Group']
)

# Create stacked bar chart
stacked_data.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 5)
)

plt.title('Stacked Earthquakes by Country and Magnitude')
plt.xlabel('Country')
plt.ylabel('Number of Earthquakes')
plt.xticks(rotation=45)
plt.legend(title='Magnitude')
plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("earthquake_1995-2023.csv")

plt.figure(figsize=(10, 6))
plt.hist(df['magnitude'], bins=20, color='steelblue', edgecolor='black')
plt.xlabel('Magnitude')
plt.ylabel('Frequency')
plt.title('Histogram of Earthquake Magnitudes')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

country_count = df['country'].value_counts().head(5)

plt.figure(figsize=(8,6))

plt.pie(
    country_count,
    autopct='%1.1f%%',
    startangle=90,
    colors=['red', 'blue', 'green', 'orange', 'purple']
)

plt.title('Top 5 Countries with Earthquakes')

# Show country names and their colors on the side
plt.legend(
    country_count.index,
    title='Countries',
    loc='center left',
    bbox_to_anchor=(1, 0.5)
)

plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.scatter(df['magnitude'], df['depth'], alpha=0.7, color='green')
plt.title('Earthquake Magnitude vs. Depth')
plt.xlabel('Magnitude')
plt.ylabel('Depth (km)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv("earthquake_1995-2023.csv")

# Convert date_time to datetime
df["date_time"] = pd.to_datetime(df["date_time"])

# Sort by date
df = df.sort_values("date_time")

# Create line plot
plt.figure(figsize=(12, 6))
plt.plot(df["date_time"], df["magnitude"])

plt.title("Earthquake Magnitude Over Time")
plt.xlabel("Date")
plt.ylabel("Magnitude")
plt.grid(True)
plt.tight_layout()
plt.show()
```
